# The flux ratio bwtween the standard N2V and constrained N2V.

To show the constrained N2V could remove the flux bias. The data we use is from the PAUS band NB645. 

```python
import pandas as pd
from pathlib import Path
from matplotlib import pyplot as plt

import numpy as np
```

```python
path = '../data saved/'
denoisedgalaxyafter = pd.read_csv(path + 'constrained_n2v_denoised.csv')
denoisedgalaxybefore = pd.read_csv(path + 'standard_n2v_denoised.csv')
```

```python
denoisedgalaxyafter = denoisedgalaxyafter[denoisedgalaxyafter['I_auto'] > 21.5]
denoisedgalaxybefore = denoisedgalaxybefore[denoisedgalaxybefore['I_auto'] > 21.5]
denoisedgalaxybefore['ratio'].hist(bins = np.linspace(-1,3,500), density = True,linestyle='--', alpha = 1, color = 'red',
                                  histtype='step', linewidth=1.5, label = 'Standard N2V')
plt.axvline(denoisedgalaxybefore['ratio'].median(), color='red', linestyle='--', alpha = 1,
            label=f'Median = {denoisedgalaxybefore['ratio'].median():.2f}')

denoisedgalaxyafter['ratio'].hist(bins = np.linspace(-1,3,500), density = True, alpha = 1, color = 'steelblue',
                                 histtype='step', linewidth=1.5, label = 'Constrained N2V')
plt.axvline(denoisedgalaxyafter['ratio'].median(), color='steelblue', alpha = 1,
            label=f'Median = {denoisedgalaxyafter['ratio'].median():.2f}')

plt.xlabel(r'$R_f$', fontsize=18)
plt.ylabel("Density", fontsize=18)
plt.legend(fontsize=11)
plt.tick_params(axis='both', which='major', labelsize=12)
```
